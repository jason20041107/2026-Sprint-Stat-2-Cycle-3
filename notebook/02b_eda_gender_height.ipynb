## Exploratory Data Analysis

本研究在完成資料清理後，先進行探索性資料分析，以了解男性與女性學生身高資料的分布情形。EDA 的主要目的，是在正式進行 Welch two-sample t-test 之前，先觀察兩組資料是否適合進行平均數比較，並確認是否有明顯的分布差異或離群值。

首先，從身高箱型圖可以看出，男性學生的身高分布整體高於女性學生。男性組的中位數與主要分布範圍皆高於女性組，表示男性學生在此資料集中普遍較高。同時，箱型圖也顯示兩組都存在少數可能的離群值，但大部分資料仍集中在合理範圍內，沒有出現明顯不合理的極端資料。

其次，從男女身高直方圖可以觀察到，女性與男性的身高分布皆呈現大致集中的型態。女性學生的身高多集中在較低的區間，而男性學生的身高分布則明顯往較高的區間移動。兩組分布雖然有部分重疊，但整體位置有所不同，顯示性別之間可能存在平均身高差異。

最後，從男女平均身高長條圖可以更直觀看出，男性學生的平均身高高於女性學生。此結果與箱型圖及直方圖的觀察一致，皆顯示男性學生在平均身高上較高。

綜合以上 EDA 結果，男性與女性學生的身高分布呈現明顯差異，且男性學生的平均身高看起來高於女性學生。因此，接下來適合使用 Welch two-sample t-test 進一步檢定男性與女性學生的平均身高是否具有統計上顯著差異。
